```python
import os
import json
import yaml
import random
import warnings
from pathlib import Path

import numpy as np
import pandas as pd
import joblib
import tensorflow as tf

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Input, Dense, Dropout
from tensorflow.keras.callbacks import EarlyStopping

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import (
    OneHotEncoder,
    StandardScaler
)
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

warnings.filterwarnings("ignore")

# ============================================================
# 1. PATH CONFIGURATION
# ============================================================

BASE_DIR = Path(
    r"C:\Users\sagni\Downloads"
    r"\AI-Based Agricultural Commodity Price Prediction"
    r" and Optimal Market Recommendation"
)

DATASET_PATH = BASE_DIR / (
    "9ef84268-d588-465a-a308-a864a43d0070.csv"
)

BASE_DIR.mkdir(parents=True, exist_ok=True)

SEED = 42
random.seed(SEED)
np.random.seed(SEED)
tf.random.set_seed(SEED)

print("=" * 65)
print("AGRICULTURAL COMMODITY PRICE PREDICTION")
print("=" * 65)

print("Dataset:", DATASET_PATH)
print("Output directory:", BASE_DIR)

if not DATASET_PATH.exists():
    raise FileNotFoundError(
        f"Dataset not found: {DATASET_PATH}"
    )

# ============================================================
# 2. LOAD DATASET
# ============================================================

df = pd.read_csv(DATASET_PATH)

print("\nOriginal dataset shape:", df.shape)
print("\nDataset columns:")
print(df.columns.tolist())

print("\nFirst five records:")
print(df.head())

# ============================================================
# 3. DATA PREPROCESSING
# ============================================================

df.columns = df.columns.str.strip()

df = df.rename(columns={
    "Min_x0020_Price": "Min_Price",
    "Max_x0020_Price": "Max_Price",
    "Modal_x0020_Price": "Modal_Price"
})

required_columns = [
    "State",
    "District",
    "Market",
    "Commodity",
    "Variety",
    "Grade",
    "Arrival_Date",
    "Min_Price",
    "Max_Price",
    "Modal_Price"
]

missing = [
    c for c in required_columns
    if c not in df.columns
]

if missing:
    raise ValueError(
        f"Missing required columns: {missing}"
    )

categorical_features = [
    "State",
    "District",
    "Market",
    "Commodity",
    "Variety",
    "Grade"
]

for col in categorical_features:
    df[col] = (
        df[col]
        .fillna("Unknown")
        .astype(str)
        .str.strip()
        .replace("", "Unknown")
    )

numeric_columns = [
    "Min_Price",
    "Max_Price",
    "Modal_Price"
]

for col in numeric_columns:
    df[col] = pd.to_numeric(
        df[col],
        errors="coerce"
    )

df["Arrival_Date"] = pd.to_datetime(
    df["Arrival_Date"],
    format="%d/%m/%Y",
    errors="coerce"
)

df = df.dropna(
    subset=["Modal_Price", "Arrival_Date"]
)

df = df[
    (df["Modal_Price"] > 0)
].copy()

df = df.drop_duplicates().reset_index(drop=True)

print("\nCleaned dataset shape:", df.shape)

print("\nMissing values:")
print(df.isnull().sum())

print("\nUnique dates:", df["Arrival_Date"].nunique())

# ============================================================
# 4. FEATURE SELECTION
# ============================================================

# Do not use Min_Price or Max_Price as inputs.
# They are reported alongside Modal_Price and
# could create target leakage.

FEATURES = categorical_features
TARGET = "Modal_Price"

X = df[FEATURES].copy()
y = df[TARGET].astype(float)

print("\nFeatures:", FEATURES)
print("Target:", TARGET)

# ============================================================
# 5. TRAIN / VALIDATION / TEST SPLIT
# ============================================================

X_train_full, X_test, y_train_full, y_test = (
    train_test_split(
        X,
        y,
        test_size=0.20,
        random_state=SEED
    )
)

X_train, X_val, y_train, y_val = (
    train_test_split(
        X_train_full,
        y_train_full,
        test_size=0.20,
        random_state=SEED
    )
)

print("\nTraining records:", len(X_train))
print("Validation records:", len(X_val))
print("Testing records:", len(X_test))

# ============================================================
# 6. ENCODING
# ============================================================

# Limit very rare categorical levels to avoid
# excessively large neural-network inputs.

encoder = OneHotEncoder(
    handle_unknown="ignore",
    min_frequency=10,
    max_categories=80,
    sparse_output=False
)

preprocessor = ColumnTransformer(
    transformers=[
        ("categorical", encoder, FEATURES)
    ],
    remainder="drop"
)

X_train_encoded = preprocessor.fit_transform(
    X_train
).astype(np.float32)

X_val_encoded = preprocessor.transform(
    X_val
).astype(np.float32)

X_test_encoded = preprocessor.transform(
    X_test
).astype(np.float32)

print(
    "\nEncoded feature count:",
    X_train_encoded.shape[1]
)

# ============================================================
# 7. RANDOM FOREST MODEL
# ============================================================

print("\nTraining Random Forest...")

rf_model = RandomForestRegressor(
    n_estimators=150,
    max_depth=20,
    min_samples_leaf=2,
    random_state=SEED,
    n_jobs=-1
)

rf_model.fit(
    X_train_encoded,
    y_train
)

rf_predictions = rf_model.predict(
    X_test_encoded
)

print("Random Forest training completed.")

# ============================================================
# 8. DEEP LEARNING MODEL
# ============================================================

print("\nTraining Neural Network...")

# Scale target for stable neural-network training.

target_scaler = StandardScaler()

y_train_scaled = target_scaler.fit_transform(
    y_train.to_numpy().reshape(-1, 1)
)

y_val_scaled = target_scaler.transform(
    y_val.to_numpy().reshape(-1, 1)
)

input_dimension = X_train_encoded.shape[1]

nn_model = Sequential([
    Input(shape=(input_dimension,)),

    Dense(256, activation="relu"),
    Dropout(0.20),

    Dense(128, activation="relu"),
    Dropout(0.15),

    Dense(64, activation="relu"),

    Dense(1, activation="linear")
])

nn_model.compile(
    optimizer=tf.keras.optimizers.Adam(
        learning_rate=0.001
    ),
    loss="mse",
    metrics=["mae"]
)

early_stopping = EarlyStopping(
    monitor="val_loss",
    patience=12,
    restore_best_weights=True
)

history = nn_model.fit(
    X_train_encoded,
    y_train_scaled,
    validation_data=(
        X_val_encoded,
        y_val_scaled
    ),
    epochs=100,
    batch_size=64,
    callbacks=[early_stopping],
    verbose=1
)

nn_predictions_scaled = nn_model.predict(
    X_test_encoded,
    verbose=0
)

nn_predictions = (
    target_scaler.inverse_transform(
        nn_predictions_scaled
    ).flatten()
)

nn_predictions = np.maximum(
    nn_predictions,
    0
)

print("Neural Network training completed.")

# ============================================================
# 9. MODEL EVALUATION
# ============================================================

def calculate_metrics(y_true, y_pred):

    mae = mean_absolute_error(
        y_true, y_pred
    )

    rmse = np.sqrt(
        mean_squared_error(
            y_true, y_pred
        )
    )

    r2 = r2_score(
        y_true, y_pred
    )

    nonzero = np.asarray(y_true) != 0

    if np.any(nonzero):
        mape = np.mean(
            np.abs(
                (
                    np.asarray(y_true)[nonzero]
                    - np.asarray(y_pred)[nonzero]
                )
                / np.asarray(y_true)[nonzero]
            )
        ) * 100
    else:
        mape = None

    return {
        "MAE": float(mae),
        "RMSE": float(rmse),
        "R2": float(r2),
        "MAPE": (
            float(mape)
            if mape is not None
            else None
        )
    }


rf_metrics = calculate_metrics(
    y_test,
    rf_predictions
)

nn_metrics = calculate_metrics(
    y_test,
    nn_predictions
)

# Baseline: predict median training price

baseline_predictions = np.full(
    len(y_test),
    y_train.median()
)

baseline_metrics = calculate_metrics(
    y_test,
    baseline_predictions
)

all_metrics = {
    "Baseline": baseline_metrics,
    "Random_Forest": rf_metrics,
    "Neural_Network": nn_metrics
}

print("\nMODEL EVALUATION RESULTS")

for model_name, metrics in all_metrics.items():

    print("\n", model_name)

    for metric, value in metrics.items():
        print(metric, ":", value)

# ============================================================
# 10. SELECT BEST MODEL
# ============================================================

candidate_metrics = {
    "Random_Forest": rf_metrics,
    "Neural_Network": nn_metrics
}

best_model_name = min(
    candidate_metrics,
    key=lambda name: candidate_metrics[name]["RMSE"]
)

print("\nBest model:", best_model_name)

if best_model_name == "Random_Forest":
    best_predictions = rf_predictions
else:
    best_predictions = nn_predictions

# ============================================================
# 11. SAVE PKL MODEL
# ============================================================

# Complete inference pipeline:
# accepts original categorical feature columns.

rf_pipeline = Pipeline([
    ("preprocessor", preprocessor),
    ("model", rf_model)
])

pkl_path = BASE_DIR / "commodity_price_model.pkl"

joblib.dump(
    rf_pipeline,
    pkl_path
)

print("\nSaved:", pkl_path)

# Save preprocessing objects for neural network

preprocessor_path = (
    BASE_DIR / "nn_preprocessor.pkl"
)

joblib.dump(
    {
        "preprocessor": preprocessor,
        "target_scaler": target_scaler,
        "features": FEATURES
    },
    preprocessor_path
)

print("Saved:", preprocessor_path)

# ============================================================
# 12. SAVE H5 MODEL
# ============================================================

h5_path = BASE_DIR / "commodity_price_model.h5"

nn_model.save(str(h5_path))

print("Saved:", h5_path)

# ============================================================
# 13. SAVE YAML CONFIGURATION
# ============================================================

config = {
    "project": (
        "AI-Based Agricultural Commodity "
        "Price Prediction and Optimal "
        "Market Recommendation"
    ),

    "dataset": str(DATASET_PATH),

    "target": TARGET,

    "features": FEATURES,

    "training": {
        "random_state": SEED,
        "test_size": 0.20,
        "validation_fraction_of_training": 0.20,
        "epochs": 100,
        "batch_size": 64
    },

    "models": {
        "random_forest": {
            "n_estimators": 150,
            "max_depth": 20,
            "min_samples_leaf": 2
        },

        "neural_network": {
            "hidden_layers": [256, 128, 64],
            "optimizer": "Adam",
            "loss": "MSE"
        }
    },

    "best_model": best_model_name,

    "output_files": {
        "h5": "commodity_price_model.h5",
        "pkl": "commodity_price_model.pkl",
        "nn_preprocessor": "nn_preprocessor.pkl",
        "json": "model_metrics.json",
        "recommendations": "market_recommendations.csv"
    }
}

yaml_path = BASE_DIR / "config.yaml"

with open(
    yaml_path,
    "w",
    encoding="utf-8"
) as file:

    yaml.safe_dump(
        config,
        file,
        sort_keys=False,
        allow_unicode=True
    )

print("Saved:", yaml_path)

# ============================================================
# 14. SAVE JSON RESULTS
# ============================================================

results = {
    "project": (
        "Agricultural Commodity Price Prediction"
    ),

    "dataset_information": {
        "total_records": int(len(df)),
        "features": FEATURES,
        "target": TARGET,
        "unique_states": int(
            df["State"].nunique()
        ),
        "unique_markets": int(
            df["Market"].nunique()
        ),
        "unique_commodities": int(
            df["Commodity"].nunique()
        ),
        "unique_dates": int(
            df["Arrival_Date"].nunique()
        )
    },

    "model_performance": all_metrics,

    "best_model": best_model_name,

    "evaluation_note": (
        "Random held-out record evaluation on "
        "a single-date market dataset. "
        "Not a future-date forecasting evaluation."
    )
}

json_path = BASE_DIR / "model_metrics.json"

with open(
    json_path,
    "w",
    encoding="utf-8"
) as file:

    json.dump(
        results,
        file,
        indent=4,
        ensure_ascii=False
    )

print("Saved:", json_path)

# ============================================================
# 15. SAVE PREDICTION CSV
# ============================================================

prediction_df = X_test.copy()

prediction_df["Actual_Price"] = (
    y_test.to_numpy()
)

prediction_df["RF_Predicted_Price"] = (
    rf_predictions
)

prediction_df["NN_Predicted_Price"] = (
    nn_predictions
)

prediction_df["Best_Predicted_Price"] = (
    best_predictions
)

prediction_df["Absolute_Error"] = np.abs(
    prediction_df["Actual_Price"]
    - prediction_df["Best_Predicted_Price"]
)

prediction_df["Percentage_Error"] = (
    prediction_df["Absolute_Error"]
    / prediction_df["Actual_Price"]
    * 100
)

prediction_path = (
    BASE_DIR / "prediction.csv"
)

prediction_df.to_csv(
    prediction_path,
    index=False
)

print("Saved:", prediction_path)

# ============================================================
# 16. MARKET RECOMMENDATION SYSTEM
# ============================================================

# Recommendations based on reported modal prices.
# These are not net-profit rankings because transport
# costs, market fees and quantities are unavailable.

market_summary = (
    df.groupby(
        [
            "State",
            "District",
            "Market",
            "Commodity",
            "Variety",
            "Grade"
        ],
        as_index=False
    )
    .agg(
        Average_Modal_Price=(
            "Modal_Price", "mean"
        ),
        Minimum_Reported_Price=(
            "Min_Price", "min"
        ),
        Maximum_Reported_Price=(
            "Max_Price", "max"
        ),
        Number_of_Records=(
            "Modal_Price", "size"
        )
    )
)

# Rank within the same commodity, variety and grade.

market_summary["Market_Rank"] = (
    market_summary.groupby(
        ["Commodity", "Variety", "Grade"]
    )["Average_Modal_Price"]
    .rank(
        ascending=False,
        method="dense"
    )
)

market_summary = market_summary.sort_values(
    [
        "Commodity",
        "Variety",
        "Grade",
        "Market_Rank"
    ]
)

recommendation_path = (
    BASE_DIR / "market_recommendations.csv"
)

market_summary.to_csv(
    recommendation_path,
    index=False
)

print("Saved:", recommendation_path)

# ============================================================
# 17. MARKET RECOMMENDATION FUNCTION
# ============================================================

def recommend_markets(
    commodity,
    variety=None,
    grade=None,
    top_n=5
):

    filtered = market_summary[
        market_summary["Commodity"].str.lower()
        == commodity.lower()
    ].copy()

    if variety is not None:
        filtered = filtered[
            filtered["Variety"].str.lower()
            == variety.lower()
        ]

    if grade is not None:
        filtered = filtered[
            filtered["Grade"].str.lower()
            == grade.lower()
        ]

    if filtered.empty:
        print(
            "No matching market records found."
        )
        return filtered

    filtered = filtered.sort_values(
        "Average_Modal_Price",
        ascending=False
    )

    return filtered.head(top_n)


# ============================================================
# 18. DEMONSTRATE RECOMMENDATIONS
# ============================================================

sample_commodity = df["Commodity"].iloc[0]

print(
    "\nSample commodity:",
    sample_commodity
)

recommended = recommend_markets(
    sample_commodity,
    top_n=5
)

print("\nTOP MARKETS")
print(recommended.to_string(index=False))

# ============================================================
# 19. SAVE DATASET SUMMARY
# ============================================================

summary = {
    "total_records": int(len(df)),
    "total_states": int(
        df["State"].nunique()
    ),
    "total_districts": int(
        df["District"].nunique()
    ),
    "total_markets": int(
        df["Market"].nunique()
    ),
    "total_commodities": int(
        df["Commodity"].nunique()
    ),
    "average_modal_price": float(
        df["Modal_Price"].mean()
    ),
    "median_modal_price": float(
        df["Modal_Price"].median()
    ),
    "minimum_modal_price": float(
        df["Modal_Price"].min()
    ),
    "maximum_modal_price": float(
        df["Modal_Price"].max()
    )
}

summary_path = (
    BASE_DIR / "dataset_summary.json"
)

with open(
    summary_path,
    "w",
    encoding="utf-8"
) as file:

    json.dump(
        summary,
        file,
        indent=4
    )

print("Saved:", summary_path)

# ============================================================
# 20. FINAL OUTPUT
# ============================================================

print("\n" + "=" * 65)
print("PROJECT EXECUTION COMPLETED")
print("=" * 65)

print("\nBest Model:", best_model_name)

print("\nGenerated Files:")

for file in sorted(BASE_DIR.iterdir()):
    if file.is_file():
        print(" -", file.name)

print("\nAll results saved successfully.")
```
