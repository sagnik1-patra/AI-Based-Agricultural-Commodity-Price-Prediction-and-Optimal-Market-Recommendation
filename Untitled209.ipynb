```python
import json
import random
import warnings
from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import joblib
import yaml
import h5py

from sklearn.model_selection import (
    train_test_split,
    KFold,
    cross_val_score
)

from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.ensemble import RandomForestRegressor

from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

warnings.filterwarnings("ignore")

# ============================================================
# 1. CONFIGURATION
# ============================================================

BASE_DIR = Path(
    r"C:\Users\sagni\Downloads"
    r"\AI-Based Agricultural Commodity Price Prediction"
    r" and Optimal Market Recommendation"
)

DATASET_PATH = BASE_DIR / (
    "9ef84268-d588-465a-a308-a864a43d0070.csv"
)

BASE_DIR.mkdir(parents=True, exist_ok=True)

SEED = 42

random.seed(SEED)
np.random.seed(SEED)

rng = np.random.default_rng(SEED)

# PSO settings

N_PARTICLES = 8
N_ITERATIONS = 8

INERTIA = 0.7
COGNITIVE = 1.5
SOCIAL = 1.5

CV_FOLDS = 3

sns.set_theme(style="whitegrid")

print("=" * 70)
print("PSO AGRICULTURAL COMMODITY PRICE PREDICTION")
print("=" * 70)

if not DATASET_PATH.exists():
    raise FileNotFoundError(
        f"Dataset not found:\n{DATASET_PATH}"
    )

# ============================================================
# 2. LOAD DATASET
# ============================================================

df = pd.read_csv(DATASET_PATH)

df.columns = df.columns.str.strip()

df = df.rename(columns={
    "Min_x0020_Price": "Min_Price",
    "Max_x0020_Price": "Max_Price",
    "Modal_x0020_Price": "Modal_Price"
})

required_columns = [
    "State",
    "District",
    "Market",
    "Commodity",
    "Variety",
    "Grade",
    "Arrival_Date",
    "Min_Price",
    "Max_Price",
    "Modal_Price"
]

missing = [
    col for col in required_columns
    if col not in df.columns
]

if missing:
    raise ValueError(
        f"Missing columns: {missing}"
    )

print("\nOriginal dataset shape:", df.shape)

# ============================================================
# 3. DATA CLEANING
# ============================================================

categorical_features = [
    "State",
    "District",
    "Market",
    "Commodity",
    "Variety",
    "Grade"
]

for col in categorical_features:

    df[col] = (
        df[col]
        .fillna("Unknown")
        .astype(str)
        .str.strip()
        .replace("", "Unknown")
    )

for col in [
    "Min_Price",
    "Max_Price",
    "Modal_Price"
]:

    df[col] = pd.to_numeric(
        df[col],
        errors="coerce"
    )

df["Arrival_Date"] = pd.to_datetime(
    df["Arrival_Date"],
    dayfirst=True,
    errors="coerce"
)

df = df.dropna(
    subset=[
        "Modal_Price",
        "Arrival_Date"
    ]
)

df = df[
    df["Modal_Price"] > 0
].copy()

df = df.drop_duplicates().reset_index(drop=True)

print("Cleaned dataset shape:", df.shape)

print(
    "Unique dates:",
    df["Arrival_Date"].nunique()
)

# ============================================================
# 4. FEATURES AND TARGET
# ============================================================

FEATURES = categorical_features
TARGET = "Modal_Price"

# Avoid target leakage:
# Min_Price and Max_Price are excluded from
# model inputs.

X = df[FEATURES].copy()
y = df[TARGET].astype(float)

# ============================================================
# 5. TRAIN / VALIDATION / TEST SPLIT
# ============================================================

X_train_full, X_test, y_train_full, y_test = (
    train_test_split(
        X,
        y,
        test_size=0.20,
        random_state=SEED
    )
)

X_train, X_val, y_train, y_val = (
    train_test_split(
        X_train_full,
        y_train_full,
        test_size=0.20,
        random_state=SEED
    )
)

print("\nTraining:", len(X_train))
print("Validation:", len(X_val))
print("Testing:", len(X_test))

# ============================================================
# 6. PREPROCESSOR
# ============================================================

def create_preprocessor():

    return ColumnTransformer(
        transformers=[
            (
                "categorical",
                OneHotEncoder(
                    handle_unknown="ignore",
                    min_frequency=10,
                    max_categories=80
                ),
                FEATURES
            )
        ]
    )

# ============================================================
# 7. PSO SEARCH SPACE
# ============================================================

# Particle dimensions:
#
# 0: n_estimators
# 1: max_depth
# 2: min_samples_split
# 3: min_samples_leaf
# 4: max_features
#
# Bounds match the AIS experiment.

LOWER_BOUNDS = np.array([
    50,
    5,
    2,
    1,
    0.40
], dtype=float)

UPPER_BOUNDS = np.array([
    180,
    30,
    15,
    8,
    1.00
], dtype=float)

DIMENSIONS = len(LOWER_BOUNDS)


def decode_particle(particle):

    return {
        "n_estimators": int(
            np.clip(
                round(particle[0]),
                50, 180
            )
        ),

        "max_depth": int(
            np.clip(
                round(particle[1]),
                5, 30
            )
        ),

        "min_samples_split": int(
            np.clip(
                round(particle[2]),
                2, 15
            )
        ),

        "min_samples_leaf": int(
            np.clip(
                round(particle[3]),
                1, 8
            )
        ),

        "max_features": float(
            np.clip(
                particle[4],
                0.40, 1.00
            )
        )
    }


# ============================================================
# 8. PSO FITNESS FUNCTION
# ============================================================

fitness_cache = {}

cv = KFold(
    n_splits=CV_FOLDS,
    shuffle=True,
    random_state=SEED
)


def evaluate_particle(particle):

    params = decode_particle(particle)

    # Rounded continuous parameter for cache reuse.
    cache_params = params.copy()

    cache_params["max_features"] = round(
        cache_params["max_features"],
        5
    )

    key = tuple(
        sorted(cache_params.items())
    )

    if key in fitness_cache:
        return fitness_cache[key]

    model = Pipeline([
        (
            "preprocessor",
            create_preprocessor()
        ),

        (
            "regressor",
            RandomForestRegressor(
                **params,
                random_state=SEED,
                n_jobs=-1
            )
        )
    ])

    scores = cross_val_score(
        model,
        X_train,
        y_train,
        cv=cv,
        scoring="neg_root_mean_squared_error",
        n_jobs=1
    )

    rmse = -float(
        np.mean(scores)
    )

    fitness_cache[key] = rmse

    return rmse


# ============================================================
# 9. INITIALIZE PSO SWARM
# ============================================================

print("\nInitializing PSO swarm...")

positions = rng.uniform(
    LOWER_BOUNDS,
    UPPER_BOUNDS,
    size=(N_PARTICLES, DIMENSIONS)
)

velocities = rng.uniform(
    -0.10,
    0.10,
    size=(N_PARTICLES, DIMENSIONS)
) * (
    UPPER_BOUNDS - LOWER_BOUNDS
)

personal_best_positions = positions.copy()

personal_best_scores = np.full(
    N_PARTICLES,
    np.inf
)

global_best_position = None
global_best_score = np.inf

history_records = []

# ============================================================
# 10. PSO OPTIMIZATION
# ============================================================

print("\nStarting PSO optimization...")

for iteration in range(
    1,
    N_ITERATIONS + 1
):

    current_scores = []

    # Evaluate particles

    for i in range(N_PARTICLES):

        score = evaluate_particle(
            positions[i]
        )

        current_scores.append(score)

        # Update personal best

        if score < personal_best_scores[i]:

            personal_best_scores[i] = score

            personal_best_positions[i] = (
                positions[i].copy()
            )

        # Update global best

        if score < global_best_score:

            global_best_score = score

            global_best_position = (
                positions[i].copy()
            )

    current_scores = np.array(
        current_scores
    )

    history_records.append({
        "Iteration": iteration,
        "Iteration_Best_RMSE": float(
            np.min(current_scores)
        ),
        "Iteration_Mean_RMSE": float(
            np.mean(current_scores)
        ),
        "Global_Best_RMSE": float(
            global_best_score
        ),
        "Unique_Evaluations": len(
            fitness_cache
        )
    })

    print(
        f"Iteration {iteration}/{N_ITERATIONS}"
        f" | Global Best CV RMSE: "
        f"{global_best_score:.4f}"
    )

    # Update particle velocities and positions

    for i in range(N_PARTICLES):

        r1 = rng.random(DIMENSIONS)
        r2 = rng.random(DIMENSIONS)

        cognitive_component = (
            COGNITIVE
            * r1
            * (
                personal_best_positions[i]
                - positions[i]
            )
        )

        social_component = (
            SOCIAL
            * r2
            * (
                global_best_position
                - positions[i]
            )
        )

        velocities[i] = (
            INERTIA * velocities[i]
            + cognitive_component
            + social_component
        )

        # Velocity clamping
        max_velocity = (
            UPPER_BOUNDS - LOWER_BOUNDS
        ) * 0.20

        velocities[i] = np.clip(
            velocities[i],
            -max_velocity,
            max_velocity
        )

        positions[i] += velocities[i]

        # Boundary handling
        positions[i] = np.clip(
            positions[i],
            LOWER_BOUNDS,
            UPPER_BOUNDS
        )

best_params = decode_particle(
    global_best_position
)

print("\nPSO OPTIMIZATION COMPLETED")
print("Best parameters:", best_params)
print("Best CV RMSE:", global_best_score)

# ============================================================
# 11. SAVE PSO OPTIMIZATION HISTORY
# ============================================================

history_df = pd.DataFrame(
    history_records
)

history_df.to_csv(
    BASE_DIR / "pso_optimization_history.csv",
    index=False
)

# ============================================================
# 12. PSO CONVERGENCE GRAPH
# ============================================================

plt.figure(figsize=(11, 6))

plt.plot(
    history_df["Iteration"],
    history_df["Global_Best_RMSE"],
    marker="o",
    linewidth=2,
    color="#2563EB",
    label="Global Best CV RMSE"
)

plt.plot(
    history_df["Iteration"],
    history_df["Iteration_Mean_RMSE"],
    linestyle="--",
    color="#10B981",
    label="Swarm Mean CV RMSE"
)

plt.title(
    "PSO Optimization Convergence",
    fontsize=15,
    fontweight="bold"
)

plt.xlabel("Iteration")
plt.ylabel("Cross-Validation RMSE")

plt.legend()
plt.tight_layout()

plt.savefig(
    BASE_DIR / "pso_convergence_graph.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# ============================================================
# 13. TRAIN BASELINE RANDOM FOREST
# ============================================================

print("\nTraining baseline Random Forest...")

baseline_model = Pipeline([
    (
        "preprocessor",
        create_preprocessor()
    ),
    (
        "regressor",
        RandomForestRegressor(
            n_estimators=150,
            max_depth=20,
            min_samples_leaf=2,
            random_state=SEED,
            n_jobs=-1
        )
    )
])

baseline_model.fit(
    X_train_full,
    y_train_full
)

# ============================================================
# 14. TRAIN PSO-OPTIMIZED MODEL
# ============================================================

print("\nTraining PSO-optimized Random Forest...")

pso_model = Pipeline([
    (
        "preprocessor",
        create_preprocessor()
    ),
    (
        "regressor",
        RandomForestRegressor(
            **best_params,
            random_state=SEED,
            n_jobs=-1
        )
    )
])

pso_model.fit(
    X_train_full,
    y_train_full
)

# ============================================================
# 15. MAKE PREDICTIONS
# ============================================================

baseline_predictions = np.maximum(
    baseline_model.predict(X_test),
    0
)

pso_predictions = np.maximum(
    pso_model.predict(X_test),
    0
)

# ============================================================
# 16. EVALUATION METRICS
# ============================================================

def calculate_metrics(y_true, y_pred):

    y_true = np.asarray(
        y_true,
        dtype=float
    )

    y_pred = np.asarray(
        y_pred,
        dtype=float
    )

    mae = mean_absolute_error(
        y_true,
        y_pred
    )

    rmse = np.sqrt(
        mean_squared_error(
            y_true,
            y_pred
        )
    )

    r2 = r2_score(
        y_true,
        y_pred
    )

    valid = y_true != 0

    mape = (
        np.mean(
            np.abs(
                (y_true[valid] - y_pred[valid])
                / y_true[valid]
            )
        ) * 100
        if valid.any()
        else np.nan
    )

    return {
        "MAE": float(mae),
        "RMSE": float(rmse),
        "R2": float(r2),
        "MAPE": float(mape)
    }


baseline_metrics = calculate_metrics(
    y_test,
    baseline_predictions
)

pso_metrics = calculate_metrics(
    y_test,
    pso_predictions
)

metrics = {
    "Baseline_RF": baseline_metrics,
    "PSO_Optimized_RF": pso_metrics
}

metrics_df = pd.DataFrame(
    metrics
).T

print("\nMODEL COMPARISON")
print(metrics_df.round(4))

metrics_df.to_csv(
    BASE_DIR / "pso_model_comparison.csv"
)

with open(
    BASE_DIR / "pso_metrics.json",
    "w",
    encoding="utf-8"
) as file:

    json.dump(
        {
            "models": metrics,
            "best_pso_parameters": best_params,
            "best_cv_rmse": float(
                global_best_score
            ),
            "optimization_algorithm": "PSO",
            "evaluation": (
                "Held-out same-date market records"
            )
        },
        file,
        indent=4
    )

# ============================================================
# 17. PSO ACCURACY GRAPH (R2 SCORE)
# ============================================================

plt.figure(figsize=(10, 6))

names = [
    "Baseline RF",
    "PSO-Optimized RF"
]

r2_values = [
    baseline_metrics["R2"],
    pso_metrics["R2"]
]

bars = plt.bar(
    names,
    r2_values,
    color=["#94A3B8", "#2563EB"]
)

plt.axhline(
    0,
    color="black",
    linewidth=0.8
)

for bar, value in zip(bars, r2_values):

    plt.annotate(
        f"{value:.4f}",
        (
            bar.get_x()
            + bar.get_width() / 2,
            bar.get_height()
        ),
        xytext=(
            0,
            5 if value >= 0 else -15
        ),
        textcoords="offset points",
        ha="center"
    )

plt.title(
    "PSO Model Performance (R² Score)",
    fontweight="bold"
)

plt.ylabel("R² Score")
plt.tight_layout()

plt.savefig(
    BASE_DIR / "pso_accuracy_graph.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# ============================================================
# 18. PSO CORRELATION HEATMAP
# ============================================================

heatmap_data = df[
    [
        "Min_Price",
        "Max_Price",
        "Modal_Price"
    ]
].copy()

heatmap_data["Price_Spread"] = (
    heatmap_data["Max_Price"]
    - heatmap_data["Min_Price"]
)

correlation = heatmap_data.corr()

plt.figure(figsize=(9, 7))

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".3f",
    linewidths=0.5,
    square=True
)

plt.title(
    "Agricultural Commodity Price Correlation Heatmap",
    fontweight="bold"
)

plt.tight_layout()

plt.savefig(
    BASE_DIR / "pso_heatmap.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# ============================================================
# 19. PSO MODEL COMPARISON GRAPH
# ============================================================

fig, axes = plt.subplots(
    2, 2,
    figsize=(14, 10)
)

metric_names = [
    "MAE",
    "RMSE",
    "R2",
    "MAPE"
]

for ax, metric in zip(
    axes.flatten(),
    metric_names
):

    values = [
        baseline_metrics[metric],
        pso_metrics[metric]
    ]

    bars = ax.bar(
        names,
        values,
        color=["#94A3B8", "#2563EB"]
    )

    ax.set_title(
        f"{metric} Comparison",
        fontweight="bold"
    )

    ax.set_ylabel(metric)

    for bar, value in zip(bars, values):

        ax.annotate(
            f"{value:.3f}",
            (
                bar.get_x()
                + bar.get_width() / 2,
                bar.get_height()
            ),
            xytext=(
                0,
                5 if value >= 0 else -15
            ),
            textcoords="offset points",
            ha="center",
            fontsize=9
        )

plt.suptitle(
    "Baseline Random Forest vs PSO-Optimized Random Forest",
    fontsize=16,
    fontweight="bold"
)

plt.tight_layout()

plt.savefig(
    BASE_DIR / "pso_comparison_graph.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# ============================================================
# 20. PSO RESULT CSV
# ============================================================

result_df = X_test.copy()

result_df["Actual_Price"] = (
    y_test.to_numpy()
)

result_df["Baseline_Prediction"] = (
    baseline_predictions
)

result_df["PSO_Prediction"] = (
    pso_predictions
)

result_df["PSO_Absolute_Error"] = np.abs(
    result_df["Actual_Price"]
    - result_df["PSO_Prediction"]
)

result_df["PSO_Percentage_Error"] = (
    result_df["PSO_Absolute_Error"]
    / result_df["Actual_Price"]
    * 100
)

result_df = result_df.reset_index(
    drop=True
)

result_df.to_csv(
    BASE_DIR / "pso_result.csv",
    index=False
)

print("\nSaved pso_result.csv")

# ============================================================
# 21. PSO RESULT GRAPH
# ============================================================

plt.figure(figsize=(10, 8))

plt.scatter(
    result_df["Actual_Price"],
    result_df["PSO_Prediction"],
    alpha=0.5,
    s=22,
    color="#2563EB",
    label="PSO Predictions"
)

min_value = min(
    result_df["Actual_Price"].min(),
    result_df["PSO_Prediction"].min()
)

max_value = max(
    result_df["Actual_Price"].max(),
    result_df["PSO_Prediction"].max()
)

plt.plot(
    [min_value, max_value],
    [min_value, max_value],
    "--",
    color="red",
    linewidth=2,
    label="Perfect Prediction"
)

plt.xlabel("Actual Modal Price")
plt.ylabel("PSO Predicted Modal Price")

plt.title(
    "PSO Actual vs Predicted Commodity Prices",
    fontweight="bold"
)

plt.legend()
plt.tight_layout()

plt.savefig(
    BASE_DIR / "pso_result_graph.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# ============================================================
# 22. PSO PREDICTION CSV
# ============================================================

prediction_df = X_test.copy()

prediction_df["Actual_Price"] = (
    y_test.to_numpy()
)

prediction_df["Baseline_Prediction"] = (
    baseline_predictions
)

prediction_df["PSO_Prediction"] = (
    pso_predictions
)

prediction_df = prediction_df.reset_index(
    drop=True
)

prediction_df.to_csv(
    BASE_DIR / "pso_prediction.csv",
    index=False
)

print("\nSaved pso_prediction.csv")

# ============================================================
# 23. PSO PREDICTION GRAPH
# ============================================================

plot_df = prediction_df.sample(
    n=min(100, len(prediction_df)),
    random_state=SEED
).sort_values(
    "Actual_Price"
).reset_index(drop=True)

plt.figure(figsize=(15, 7))

plt.plot(
    plot_df.index,
    plot_df["Actual_Price"],
    label="Actual Price",
    color="black",
    linewidth=2.5
)

plt.plot(
    plot_df.index,
    plot_df["Baseline_Prediction"],
    label="Baseline RF",
    color="#94A3B8",
    linestyle="--"
)

plt.plot(
    plot_df.index,
    plot_df["PSO_Prediction"],
    label="PSO-Optimized RF",
    color="#2563EB",
    linewidth=2
)

plt.xlabel("Sorted Test Sample Index")
plt.ylabel("Modal Price")

plt.title(
    "PSO Commodity Price Prediction Comparison",
    fontweight="bold"
)

plt.legend()
plt.tight_layout()

plt.savefig(
    BASE_DIR / "pso_prediction_graph.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# ============================================================
# 24. PSO MARKET RECOMMENDATIONS
# ============================================================

# Rank markets by reported modal prices.
# This is not a net-profit recommendation.

market_df = (
    df.groupby(
        [
            "State",
            "District",
            "Market",
            "Commodity",
            "Variety",
            "Grade"
        ],
        as_index=False
    )
    .agg(
        Average_Modal_Price=(
            "Modal_Price",
            "mean"
        ),
        Min_Reported_Price=(
            "Min_Price",
            "min"
        ),
        Max_Reported_Price=(
            "Max_Price",
            "max"
        ),
        Records=(
            "Modal_Price",
            "size"
        )
    )
)

market_df["Market_Rank"] = (
    market_df.groupby(
        [
            "Commodity",
            "Variety",
            "Grade"
        ]
    )["Average_Modal_Price"]
    .rank(
        ascending=False,
        method="dense"
    )
)

market_df = market_df.sort_values(
    [
        "Commodity",
        "Variety",
        "Grade",
        "Market_Rank"
    ]
)

market_df.to_csv(
    BASE_DIR / "pso_market_recommendations.csv",
    index=False
)

print("\nSaved pso_market_recommendations.csv")

# ============================================================
# 25. MARKET RECOMMENDATION FUNCTION
# ============================================================

def pso_recommend_markets(
    commodity,
    variety=None,
    grade=None,
    top_n=5
):

    filtered = market_df[
        market_df["Commodity"].str.lower()
        == commodity.lower()
    ].copy()

    if variety is not None:

        filtered = filtered[
            filtered["Variety"].str.lower()
            == variety.lower()
        ]

    if grade is not None:

        filtered = filtered[
            filtered["Grade"].str.lower()
            == grade.lower()
        ]

    filtered = filtered.sort_values(
        "Average_Modal_Price",
        ascending=False
    )

    return filtered.head(top_n)


sample_commodity = df["Commodity"].iloc[0]

print("\nSample commodity:", sample_commodity)

print(
    pso_recommend_markets(
        sample_commodity,
        top_n=5
    ).to_string(index=False)
)

# ============================================================
# 26. SAVE PSO MODEL AS PKL
# ============================================================

joblib.dump(
    pso_model,
    BASE_DIR / "pso_model.pkl"
)

print("\nSaved pso_model.pkl")

# ============================================================
# 27. SAVE PSO METADATA AS H5
# ============================================================

# This H5 contains optimization metadata.
# The trained sklearn model is saved as PKL.

with h5py.File(
    BASE_DIR / "pso_model.h5",
    "w"
) as h5:

    h5.attrs["algorithm"] = (
        "Particle Swarm Optimization"
    )

    h5.attrs["predictor"] = (
        "RandomForestRegressor"
    )

    h5.attrs["model_file"] = (
        "pso_model.pkl"
    )

    h5.attrs["best_cv_rmse"] = (
        global_best_score
    )

    params_group = h5.create_group(
        "best_parameters"
    )

    for key, value in best_params.items():
        params_group.attrs[key] = value

    h5.create_dataset(
        "convergence_rmse",
        data=history_df[
            "Global_Best_RMSE"
        ].to_numpy()
    )

    h5.create_dataset(
        "best_particle",
        data=global_best_position
    )

print("Saved pso_model.h5")

# ============================================================
# 28. SAVE PSO YAML CONFIGURATION
# ============================================================

config = {
    "project": (
        "PSO-Based Agricultural Commodity "
        "Price Prediction"
    ),
    "dataset": str(DATASET_PATH),
    "algorithm": "PSO",
    "predictor": "RandomForestRegressor",
    "target": TARGET,
    "features": FEATURES,
    "seed": SEED,
    "pso": {
        "particles": N_PARTICLES,
        "iterations": N_ITERATIONS,
        "inertia": INERTIA,
        "cognitive": COGNITIVE,
        "social": SOCIAL,
        "cv_folds": CV_FOLDS
    },
    "best_parameters": best_params,
    "best_cv_rmse": float(
        global_best_score
    )
}

with open(
    BASE_DIR / "pso_config.yaml",
    "w",
    encoding="utf-8"
) as file:

    yaml.safe_dump(
        config,
        file,
        sort_keys=False
    )

print("Saved pso_config.yaml")

# ============================================================
# 29. SAVE PSO PROJECT SUMMARY
# ============================================================

summary = {
    "dataset_records": int(len(df)),
    "unique_states": int(
        df["State"].nunique()
    ),
    "unique_markets": int(
        df["Market"].nunique()
    ),
    "unique_commodities": int(
        df["Commodity"].nunique()
    ),
    "unique_dates": int(
        df["Arrival_Date"].nunique()
    ),
    "algorithm": "PSO",
    "best_parameters": best_params,
    "baseline_metrics": baseline_metrics,
    "pso_metrics": pso_metrics,
    "best_cv_rmse": float(
        global_best_score
    ),
    "evaluation_note": (
        "Test metrics evaluate held-out market "
        "records, not future-date forecasting."
    )
}

with open(
    BASE_DIR / "pso_project_summary.json",
    "w",
    encoding="utf-8"
) as file:

    json.dump(
        summary,
        file,
        indent=4
    )

# ============================================================
# 30. PSO IMPROVEMENT ANALYSIS
# ============================================================

baseline_rmse = baseline_metrics["RMSE"]
pso_rmse = pso_metrics["RMSE"]

improvement = (
    (baseline_rmse - pso_rmse)
    / baseline_rmse
) * 100 if baseline_rmse > 0 else np.nan

print("\n" + "=" * 70)
print("PSO OPTIMIZATION RESULTS")
print("=" * 70)

print("\nBaseline RMSE:", baseline_rmse)
print("PSO RMSE:", pso_rmse)

print(
    "PSO RMSE Improvement (%):",
    round(improvement, 4)
)

if improvement > 0:
    print("PSO improved test RMSE.")
elif improvement < 0:
    print("PSO did not improve test RMSE.")
else:
    print("PSO and baseline RMSE are equal.")

# ============================================================
# 31. FINAL FILE LIST
# ============================================================

print("\n" + "=" * 70)
print("ALL PSO RESULTS GENERATED")
print("=" * 70)

for file in sorted(BASE_DIR.glob("pso_*")):

    if file.is_file():
        print(" -", file.name)

print("\nOutput directory:")
print(BASE_DIR)

print("\nPSO PROJECT COMPLETED SUCCESSFULLY")
```
