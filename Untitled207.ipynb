```python
import os
import json
import random
import warnings
from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import joblib
import yaml
import tensorflow as tf

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Input, Dense, Dropout
from tensorflow.keras.callbacks import EarlyStopping

warnings.filterwarnings("ignore")

# ============================================================
# 1. PROJECT CONFIGURATION
# ============================================================

BASE_DIR = Path(
    r"C:\Users\sagni\Downloads"
    r"\AI-Based Agricultural Commodity Price Prediction"
    r" and Optimal Market Recommendation"
)

DATASET_PATH = BASE_DIR / (
    "9ef84268-d588-465a-a308-a864a43d0070.csv"
)

BASE_DIR.mkdir(parents=True, exist_ok=True)

SEED = 42

random.seed(SEED)
np.random.seed(SEED)
tf.random.set_seed(SEED)

sns.set_theme(style="whitegrid")

plt.rcParams["figure.figsize"] = (12, 6)
plt.rcParams["figure.dpi"] = 120

print("=" * 70)
print("AGRICULTURAL COMMODITY PRICE ANALYSIS")
print("=" * 70)

if not DATASET_PATH.exists():
    raise FileNotFoundError(
        f"Dataset not found:\n{DATASET_PATH}"
    )

# ============================================================
# 2. LOAD DATASET
# ============================================================

df = pd.read_csv(DATASET_PATH)

df.columns = df.columns.str.strip()

df = df.rename(columns={
    "Min_x0020_Price": "Min_Price",
    "Max_x0020_Price": "Max_Price",
    "Modal_x0020_Price": "Modal_Price"
})

print("\nOriginal dataset shape:", df.shape)

required_columns = [
    "State",
    "District",
    "Market",
    "Commodity",
    "Variety",
    "Grade",
    "Arrival_Date",
    "Min_Price",
    "Max_Price",
    "Modal_Price"
]

missing = [
    col for col in required_columns
    if col not in df.columns
]

if missing:
    raise ValueError(
        f"Missing required columns: {missing}"
    )

# ============================================================
# 3. DATA CLEANING
# ============================================================

categorical_columns = [
    "State",
    "District",
    "Market",
    "Commodity",
    "Variety",
    "Grade"
]

numeric_columns = [
    "Min_Price",
    "Max_Price",
    "Modal_Price"
]

for col in categorical_columns:
    df[col] = (
        df[col]
        .fillna("Unknown")
        .astype(str)
        .str.strip()
        .replace("", "Unknown")
    )

for col in numeric_columns:
    df[col] = pd.to_numeric(
        df[col],
        errors="coerce"
    )

df["Arrival_Date"] = pd.to_datetime(
    df["Arrival_Date"],
    dayfirst=True,
    errors="coerce"
)

df = df.dropna(
    subset=["Modal_Price", "Arrival_Date"]
)

df = df[
    df["Modal_Price"] > 0
].copy()

df = df.drop_duplicates().reset_index(drop=True)

print("\nCleaned dataset shape:", df.shape)
print("Unique dates:", df["Arrival_Date"].nunique())

# ============================================================
# 4. FEATURE SELECTION
# ============================================================

# Avoid target leakage:
# Min_Price and Max_Price are NOT model inputs.

FEATURES = categorical_columns
TARGET = "Modal_Price"

X = df[FEATURES].copy()
y = df[TARGET].astype(float)

# ============================================================
# 5. TRAIN / VALIDATION / TEST SPLIT
# ============================================================

X_train_full, X_test, y_train_full, y_test = (
    train_test_split(
        X,
        y,
        test_size=0.20,
        random_state=SEED
    )
)

X_train, X_val, y_train, y_val = (
    train_test_split(
        X_train_full,
        y_train_full,
        test_size=0.20,
        random_state=SEED
    )
)

print("\nTraining:", len(X_train))
print("Validation:", len(X_val))
print("Testing:", len(X_test))

# ============================================================
# 6. PREPROCESSING
# ============================================================

encoder = OneHotEncoder(
    handle_unknown="ignore",
    min_frequency=10,
    max_categories=80,
    sparse_output=False
)

preprocessor = ColumnTransformer(
    transformers=[
        ("categorical", encoder, FEATURES)
    ]
)

X_train_encoded = preprocessor.fit_transform(
    X_train
).astype(np.float32)

X_val_encoded = preprocessor.transform(
    X_val
).astype(np.float32)

X_test_encoded = preprocessor.transform(
    X_test
).astype(np.float32)

print(
    "\nEncoded features:",
    X_train_encoded.shape[1]
)

# ============================================================
# 7. RANDOM FOREST MODEL
# ============================================================

print("\nTraining Random Forest...")

rf_model = RandomForestRegressor(
    n_estimators=150,
    max_depth=20,
    min_samples_leaf=2,
    random_state=SEED,
    n_jobs=-1
)

rf_model.fit(
    X_train_encoded,
    y_train
)

rf_predictions = rf_model.predict(
    X_test_encoded
)

rf_predictions = np.maximum(
    rf_predictions, 0
)

# ============================================================
# 8. NEURAL NETWORK MODEL
# ============================================================

print("\nTraining Neural Network...")

target_scaler = StandardScaler()

y_train_scaled = target_scaler.fit_transform(
    y_train.to_numpy().reshape(-1, 1)
)

y_val_scaled = target_scaler.transform(
    y_val.to_numpy().reshape(-1, 1)
)

nn_model = Sequential([
    Input(shape=(X_train_encoded.shape[1],)),

    Dense(256, activation="relu"),
    Dropout(0.20),

    Dense(128, activation="relu"),
    Dropout(0.15),

    Dense(64, activation="relu"),

    Dense(1)
])

nn_model.compile(
    optimizer=tf.keras.optimizers.Adam(
        learning_rate=0.001
    ),
    loss="mse",
    metrics=["mae"]
)

early_stopping = EarlyStopping(
    monitor="val_loss",
    patience=12,
    restore_best_weights=True
)

history = nn_model.fit(
    X_train_encoded,
    y_train_scaled,
    validation_data=(
        X_val_encoded,
        y_val_scaled
    ),
    epochs=100,
    batch_size=64,
    callbacks=[early_stopping],
    verbose=1
)

nn_scaled_predictions = nn_model.predict(
    X_test_encoded,
    verbose=0
)

nn_predictions = target_scaler.inverse_transform(
    nn_scaled_predictions
).flatten()

nn_predictions = np.maximum(
    nn_predictions, 0
)

# ============================================================
# 9. BASELINE MODEL
# ============================================================

baseline_predictions = np.full(
    len(y_test),
    y_train.median()
)

# ============================================================
# 10. MODEL EVALUATION
# ============================================================

def calculate_metrics(y_true, y_pred):

    y_true = np.asarray(y_true, dtype=float)
    y_pred = np.asarray(y_pred, dtype=float)

    mae = mean_absolute_error(
        y_true, y_pred
    )

    rmse = np.sqrt(
        mean_squared_error(y_true, y_pred)
    )

    r2 = r2_score(
        y_true, y_pred
    )

    valid = y_true != 0

    mape = (
        np.mean(
            np.abs(
                (y_true[valid] - y_pred[valid])
                / y_true[valid]
            )
        ) * 100
        if valid.any()
        else np.nan
    )

    return {
        "MAE": float(mae),
        "RMSE": float(rmse),
        "R2": float(r2),
        "MAPE": float(mape)
    }


model_predictions = {
    "Baseline": baseline_predictions,
    "Random Forest": rf_predictions,
    "Neural Network": nn_predictions
}

metrics = {
    name: calculate_metrics(y_test, pred)
    for name, pred in model_predictions.items()
}

metrics_df = pd.DataFrame(metrics).T

print("\nMODEL PERFORMANCE")
print(metrics_df.round(4))

metrics_df.to_csv(
    BASE_DIR / "model_comparison.csv"
)

with open(
    BASE_DIR / "evaluation_metrics.json",
    "w",
    encoding="utf-8"
) as f:
    json.dump(metrics, f, indent=4)

# ============================================================
# 11. ACCURACY GRAPH (R2 SCORE)
# ============================================================

print("\nGenerating accuracy graph...")

fig, ax = plt.subplots(figsize=(10, 6))

names = list(metrics.keys())

r2_values = [
    metrics[name]["R2"]
    for name in names
]

bars = ax.bar(
    names,
    r2_values,
    color=["#9CA3AF", "#2563EB", "#10B981"]
)

ax.axhline(
    0,
    color="black",
    linewidth=0.8
)

ax.set_title(
    "Model Performance Comparison (R² Score)",
    fontsize=15,
    fontweight="bold"
)

ax.set_ylabel("R² Score")

for bar, value in zip(bars, r2_values):
    ax.annotate(
        f"{value:.4f}",
        xy=(
            bar.get_x() + bar.get_width() / 2,
            bar.get_height()
        ),
        xytext=(0, 5 if value >= 0 else -15),
        textcoords="offset points",
        ha="center",
        fontsize=10
    )

plt.tight_layout()

plt.savefig(
    BASE_DIR / "accuracy_graph.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# ============================================================
# 12. CORRELATION HEATMAP
# ============================================================

print("\nGenerating heatmap...")

heatmap_df = df[
    [
        "Min_Price",
        "Max_Price",
        "Modal_Price"
    ]
].copy()

heatmap_df["Price_Spread"] = (
    heatmap_df["Max_Price"]
    - heatmap_df["Min_Price"]
)

corr = heatmap_df.corr(
    method="pearson"
)

plt.figure(figsize=(9, 7))

sns.heatmap(
    corr,
    annot=True,
    cmap="coolwarm",
    fmt=".3f",
    linewidths=0.5,
    square=True
)

plt.title(
    "Agricultural Commodity Price Correlation Heatmap",
    fontsize=14,
    fontweight="bold"
)

plt.tight_layout()

plt.savefig(
    BASE_DIR / "heatmap.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# ============================================================
# 13. MODEL COMPARISON GRAPH
# ============================================================

print("\nGenerating comparison graph...")

fig, axes = plt.subplots(
    2, 2,
    figsize=(14, 10)
)

metric_names = [
    "MAE",
    "RMSE",
    "R2",
    "MAPE"
]

colors = [
    "#64748B",
    "#2563EB",
    "#10B981"
]

for ax, metric in zip(
    axes.flatten(),
    metric_names
):

    values = [
        metrics[name][metric]
        for name in names
    ]

    bars = ax.bar(
        names,
        values,
        color=colors
    )

    ax.set_title(
        f"{metric} Comparison",
        fontweight="bold"
    )

    ax.set_ylabel(metric)

    ax.tick_params(
        axis="x",
        rotation=15
    )

    for bar, value in zip(bars, values):
        ax.annotate(
            f"{value:.2f}",
            (
                bar.get_x()
                + bar.get_width() / 2,
                bar.get_height()
            ),
            xytext=(0, 4 if value >= 0 else -14),
            textcoords="offset points",
            ha="center",
            fontsize=9
        )

plt.suptitle(
    "Machine Learning Model Comparison",
    fontsize=17,
    fontweight="bold"
)

plt.tight_layout()

plt.savefig(
    BASE_DIR / "comparison_graph.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# ============================================================
# 14. SELECT BEST MODEL
# ============================================================

candidate_names = [
    "Random Forest",
    "Neural Network"
]

best_model_name = min(
    candidate_names,
    key=lambda name: metrics[name]["RMSE"]
)

best_predictions = model_predictions[
    best_model_name
]

print("\nBest model:", best_model_name)

# ============================================================
# 15. RESULT CSV
# ============================================================

result_df = X_test.copy()

result_df["Actual_Price"] = (
    y_test.to_numpy()
)

result_df["Predicted_Price"] = (
    best_predictions
)

result_df["Absolute_Error"] = np.abs(
    result_df["Actual_Price"]
    - result_df["Predicted_Price"]
)

result_df["Percentage_Error"] = (
    result_df["Absolute_Error"]
    / result_df["Actual_Price"]
    * 100
)

result_df = result_df.reset_index(drop=True)

result_df.to_csv(
    BASE_DIR / "result.csv",
    index=False
)

print("\nSaved result.csv")

# ============================================================
# 16. RESULT GRAPH
# ============================================================

print("\nGenerating result graph...")

plt.figure(figsize=(10, 8))

plt.scatter(
    result_df["Actual_Price"],
    result_df["Predicted_Price"],
    alpha=0.5,
    s=22,
    color="#2563EB"
)

min_price = min(
    result_df["Actual_Price"].min(),
    result_df["Predicted_Price"].min()
)

max_price = max(
    result_df["Actual_Price"].max(),
    result_df["Predicted_Price"].max()
)

plt.plot(
    [min_price, max_price],
    [min_price, max_price],
    "--",
    color="red",
    linewidth=2,
    label="Perfect Prediction"
)

plt.xlabel("Actual Modal Price")
plt.ylabel("Predicted Modal Price")

plt.title(
    f"Actual vs Predicted Prices ({best_model_name})",
    fontsize=15,
    fontweight="bold"
)

plt.legend()

plt.tight_layout()

plt.savefig(
    BASE_DIR / "result_graph.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# ============================================================
# 17. PREDICTION CSV
# ============================================================

prediction_df = X_test.copy()

prediction_df["Actual_Price"] = (
    y_test.to_numpy()
)

prediction_df["Baseline_Prediction"] = (
    baseline_predictions
)

prediction_df["RF_Prediction"] = (
    rf_predictions
)

prediction_df["NN_Prediction"] = (
    nn_predictions
)

prediction_df["Best_Model_Prediction"] = (
    best_predictions
)

prediction_df = prediction_df.reset_index(drop=True)

prediction_df.to_csv(
    BASE_DIR / "prediction.csv",
    index=False
)

print("\nSaved prediction.csv")

# ============================================================
# 18. PREDICTION GRAPH
# ============================================================

print("\nGenerating prediction graph...")

# Sort a sample by actual price for clear visualization.

plot_df = prediction_df.sample(
    n=min(100, len(prediction_df)),
    random_state=SEED
).sort_values(
    "Actual_Price"
).reset_index(drop=True)

plt.figure(figsize=(15, 7))

plt.plot(
    plot_df.index,
    plot_df["Actual_Price"],
    label="Actual Price",
    color="black",
    linewidth=2.5
)

plt.plot(
    plot_df.index,
    plot_df["RF_Prediction"],
    label="Random Forest",
    linestyle="--",
    color="#2563EB"
)

plt.plot(
    plot_df.index,
    plot_df["NN_Prediction"],
    label="Neural Network",
    linestyle=":",
    color="#10B981",
    linewidth=2
)

plt.xlabel("Sorted Test Sample Index")
plt.ylabel("Modal Price")

plt.title(
    "Agricultural Commodity Price Prediction Comparison",
    fontsize=15,
    fontweight="bold"
)

plt.legend()

plt.tight_layout()

plt.savefig(
    BASE_DIR / "prediction_graph.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# ============================================================
# 19. NEURAL NETWORK TRAINING GRAPH
# ============================================================

plt.figure(figsize=(11, 6))

plt.plot(
    history.history["loss"],
    label="Training Loss"
)

plt.plot(
    history.history["val_loss"],
    label="Validation Loss"
)

plt.xlabel("Epoch")
plt.ylabel("Scaled Target MSE")

plt.title(
    "Neural Network Training and Validation Loss"
)

plt.legend()

plt.tight_layout()

plt.savefig(
    BASE_DIR / "training_history.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# ============================================================
# 20. MARKET RECOMMENDATION CSV
# ============================================================

market_df = (
    df.groupby(
        [
            "State",
            "District",
            "Market",
            "Commodity",
            "Variety",
            "Grade"
        ],
        as_index=False
    )
    .agg(
        Average_Modal_Price=(
            "Modal_Price", "mean"
        ),
        Min_Reported_Price=(
            "Min_Price", "min"
        ),
        Max_Reported_Price=(
            "Max_Price", "max"
        ),
        Records=(
            "Modal_Price", "size"
        )
    )
)

market_df["Rank"] = (
    market_df.groupby(
        ["Commodity", "Variety", "Grade"]
    )["Average_Modal_Price"]
    .rank(
        ascending=False,
        method="dense"
    )
)

market_df = market_df.sort_values(
    [
        "Commodity",
        "Variety",
        "Grade",
        "Rank"
    ]
)

market_df.to_csv(
    BASE_DIR / "market_recommendations.csv",
    index=False
)

print("\nSaved market_recommendations.csv")

# ============================================================
# 21. SAVE MODELS
# ============================================================

rf_pipeline = Pipeline([
    ("preprocessor", preprocessor),
    ("model", rf_model)
])

joblib.dump(
    rf_pipeline,
    BASE_DIR / "commodity_price_model.pkl"
)

joblib.dump(
    {
        "preprocessor": preprocessor,
        "target_scaler": target_scaler,
        "features": FEATURES
    },
    BASE_DIR / "nn_preprocessor.pkl"
)

nn_model.save(
    str(BASE_DIR / "commodity_price_model.h5")
)

# ============================================================
# 22. SAVE YAML CONFIG
# ============================================================

config = {
    "project": (
        "AI-Based Agricultural Commodity Price Prediction "
        "and Optimal Market Recommendation"
    ),
    "dataset": str(DATASET_PATH),
    "target": TARGET,
    "features": FEATURES,
    "random_state": SEED,
    "test_size": 0.20,
    "best_model": best_model_name,
    "models": [
        "Baseline",
        "Random Forest",
        "Neural Network"
    ],
    "metrics": [
        "MAE",
        "RMSE",
        "R2",
        "MAPE"
    ]
}

with open(
    BASE_DIR / "config.yaml",
    "w",
    encoding="utf-8"
) as f:
    yaml.safe_dump(
        config,
        f,
        sort_keys=False
    )

# ============================================================
# 23. SAVE FINAL SUMMARY JSON
# ============================================================

summary = {
    "total_records": int(len(df)),
    "total_states": int(df["State"].nunique()),
    "total_markets": int(df["Market"].nunique()),
    "total_commodities": int(
        df["Commodity"].nunique()
    ),
    "unique_dates": int(
        df["Arrival_Date"].nunique()
    ),
    "best_model": best_model_name,
    "model_metrics": metrics,
    "note": (
        "Evaluation measures held-out market record "
        "price estimation, not future-date forecasting."
    )
}

with open(
    BASE_DIR / "project_summary.json",
    "w",
    encoding="utf-8"
) as f:
    json.dump(
        summary,
        f,
        indent=4
    )

# ============================================================
# 24. FINAL OUTPUT
# ============================================================

print("\n" + "=" * 70)
print("ALL RESULTS GENERATED SUCCESSFULLY")
print("=" * 70)

print("\nBEST MODEL:", best_model_name)

print("\nMODEL PERFORMANCE:")
print(metrics_df.round(4))

print("\nGENERATED FILES:")

for file in sorted(BASE_DIR.iterdir()):
    if file.is_file():
        print(" -", file.name)

print("\nOutput directory:")
print(BASE_DIR)

print("\nPROJECT COMPLETED")
```
