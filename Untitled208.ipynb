```python
import json
import random
import warnings
from pathlib import Path

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import joblib
import yaml
import h5py

from sklearn.model_selection import (
    train_test_split,
    KFold,
    cross_val_score
)
from sklearn.preprocessing import OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

warnings.filterwarnings("ignore")

# ============================================================
# 1. CONFIGURATION
# ============================================================

BASE_DIR = Path(
    r"C:\Users\sagni\Downloads"
    r"\AI-Based Agricultural Commodity Price Prediction"
    r" and Optimal Market Recommendation"
)

DATASET_PATH = BASE_DIR / (
    "9ef84268-d588-465a-a308-a864a43d0070.csv"
)

BASE_DIR.mkdir(parents=True, exist_ok=True)

SEED = 42
np.random.seed(SEED)
random.seed(SEED)

# AIS optimization parameters
POPULATION_SIZE = 8
GENERATIONS = 8
TOP_ANTIBODIES = 3
CLONES_PER_ANTIBODY = 3
RANDOM_NEW_ANTIBODIES = 2

CV_FOLDS = 3

sns.set_theme(style="whitegrid")

print("=" * 70)
print("AIS AGRICULTURAL COMMODITY PRICE PREDICTION")
print("=" * 70)

if not DATASET_PATH.exists():
    raise FileNotFoundError(
        f"Dataset not found: {DATASET_PATH}"
    )

# ============================================================
# 2. LOAD DATASET
# ============================================================

df = pd.read_csv(DATASET_PATH)

df.columns = df.columns.str.strip()

df = df.rename(columns={
    "Min_x0020_Price": "Min_Price",
    "Max_x0020_Price": "Max_Price",
    "Modal_x0020_Price": "Modal_Price"
})

required_columns = [
    "State",
    "District",
    "Market",
    "Commodity",
    "Variety",
    "Grade",
    "Arrival_Date",
    "Min_Price",
    "Max_Price",
    "Modal_Price"
]

missing = [
    col for col in required_columns
    if col not in df.columns
]

if missing:
    raise ValueError(
        f"Missing columns: {missing}"
    )

print("\nOriginal shape:", df.shape)

# ============================================================
# 3. DATA CLEANING
# ============================================================

categorical_features = [
    "State",
    "District",
    "Market",
    "Commodity",
    "Variety",
    "Grade"
]

for col in categorical_features:
    df[col] = (
        df[col]
        .fillna("Unknown")
        .astype(str)
        .str.strip()
        .replace("", "Unknown")
    )

for col in [
    "Min_Price",
    "Max_Price",
    "Modal_Price"
]:
    df[col] = pd.to_numeric(
        df[col],
        errors="coerce"
    )

df["Arrival_Date"] = pd.to_datetime(
    df["Arrival_Date"],
    dayfirst=True,
    errors="coerce"
)

df = df.dropna(
    subset=[
        "Modal_Price",
        "Arrival_Date"
    ]
)

df = df[
    df["Modal_Price"] > 0
].copy()

df = df.drop_duplicates().reset_index(drop=True)

print("Cleaned shape:", df.shape)
print("Unique dates:", df["Arrival_Date"].nunique())

# ============================================================
# 4. FEATURES AND TARGET
# ============================================================

FEATURES = categorical_features
TARGET = "Modal_Price"

# Avoid target leakage:
# Min_Price and Max_Price are not model inputs.

X = df[FEATURES].copy()
y = df[TARGET].astype(float)

# ============================================================
# 5. TRAIN / VALIDATION / TEST SPLIT
# ============================================================

X_train_full, X_test, y_train_full, y_test = (
    train_test_split(
        X,
        y,
        test_size=0.20,
        random_state=SEED
    )
)

X_train, X_val, y_train, y_val = (
    train_test_split(
        X_train_full,
        y_train_full,
        test_size=0.20,
        random_state=SEED
    )
)

print("\nTraining:", len(X_train))
print("Validation:", len(X_val))
print("Testing:", len(X_test))

# ============================================================
# 6. PREPROCESSOR
# ============================================================

def create_preprocessor():

    return ColumnTransformer(
        transformers=[
            (
                "categorical",
                OneHotEncoder(
                    handle_unknown="ignore",
                    min_frequency=10,
                    max_categories=80
                ),
                FEATURES
            )
        ]
    )

# ============================================================
# 7. AIS SEARCH SPACE
# ============================================================

# Antibody genes:
# [n_estimators, max_depth,
#  min_samples_split, min_samples_leaf,
#  max_features_fraction]

BOUNDS = [
    (50, 180),
    (5, 30),
    (2, 15),
    (1, 8),
    (0.40, 1.00)
]

INTEGER_GENES = [0, 1, 2, 3]

rng = np.random.default_rng(SEED)


def random_antibody():

    genes = []

    for i, (low, high) in enumerate(BOUNDS):

        if i in INTEGER_GENES:
            value = rng.integers(
                low,
                high + 1
            )
        else:
            value = rng.uniform(
                low,
                high
            )

        genes.append(float(value))

    return np.array(genes)


def decode_antibody(antibody):

    return {
        "n_estimators": int(
            round(antibody[0])
        ),
        "max_depth": int(
            round(antibody[1])
        ),
        "min_samples_split": int(
            round(antibody[2])
        ),
        "min_samples_leaf": int(
            round(antibody[3])
        ),
        "max_features": float(
            antibody[4]
        )
    }


def mutate_antibody(
    antibody,
    mutation_strength=0.20
):

    clone = antibody.copy()

    for i, (low, high) in enumerate(BOUNDS):

        if rng.random() < 0.75:

            noise = rng.normal(
                0,
                (high - low) * mutation_strength
            )

            clone[i] += noise

            clone[i] = np.clip(
                clone[i],
                low,
                high
            )

    return clone


# ============================================================
# 8. FITNESS FUNCTION
# ============================================================

fitness_cache = {}

cv = KFold(
    n_splits=CV_FOLDS,
    shuffle=True,
    random_state=SEED
)


def evaluate_antibody(antibody):

    params = decode_antibody(antibody)

    key = tuple(
        sorted(params.items())
    )

    if key in fitness_cache:
        return fitness_cache[key]

    model = Pipeline([
        (
            "preprocessor",
            create_preprocessor()
        ),
        (
            "regressor",
            RandomForestRegressor(
                **params,
                random_state=SEED,
                n_jobs=-1
            )
        )
    ])

    scores = cross_val_score(
        model,
        X_train,
        y_train,
        cv=cv,
        scoring="neg_root_mean_squared_error",
        n_jobs=1
    )

    rmse = -float(np.mean(scores))

    fitness_cache[key] = rmse

    return rmse


# ============================================================
# 9. AIS CLONAL SELECTION ALGORITHM
# ============================================================

print("\nStarting AIS optimization...")

population = [
    random_antibody()
    for _ in range(POPULATION_SIZE)
]

best_antibody = None
best_fitness = float("inf")

history_records = []

for generation in range(
    1,
    GENERATIONS + 1
):

    evaluated = [
        (
            evaluate_antibody(antibody),
            antibody
        )
        for antibody in population
    ]

    evaluated.sort(
        key=lambda item: item[0]
    )

    current_best_fitness = evaluated[0][0]

    if current_best_fitness < best_fitness:

        best_fitness = current_best_fitness

        best_antibody = (
            evaluated[0][1].copy()
        )

    # Select high-affinity antibodies
    selected = evaluated[
        :TOP_ANTIBODIES
    ]

    clones = []

    for rank, (fitness, antibody) in enumerate(
        selected
    ):

        # Higher-affinity antibodies
        # receive smaller mutation strength.

        mutation_strength = (
            0.05 + rank * 0.08
        )

        for _ in range(
            CLONES_PER_ANTIBODY
        ):

            clone = mutate_antibody(
                antibody,
                mutation_strength
            )

            clones.append(clone)

    # Evaluate cloned antibodies
    clone_evaluated = [
        (
            evaluate_antibody(clone),
            clone
        )
        for clone in clones
    ]

    combined = (
        evaluated + clone_evaluated
    )

    combined.sort(
        key=lambda item: item[0]
    )

    # Elitism + diversity injection
    keep_count = max(
        1,
        POPULATION_SIZE
        - RANDOM_NEW_ANTIBODIES
    )

    population = [
        item[1].copy()
        for item in combined[:keep_count]
    ]

    for _ in range(
        RANDOM_NEW_ANTIBODIES
    ):
        population.append(
            random_antibody()
        )

    if combined[0][0] < best_fitness:

        best_fitness = combined[0][0]

        best_antibody = (
            combined[0][1].copy()
        )

    history_records.append({
        "Generation": generation,
        "Generation_Best_RMSE": float(
            combined[0][0]
        ),
        "Global_Best_RMSE": float(
            best_fitness
        ),
        "Evaluations_Cached": len(
            fitness_cache
        )
    })

    print(
        f"Generation {generation}/{GENERATIONS}"
        f" | Best CV RMSE: {best_fitness:.4f}"
    )

best_params = decode_antibody(
    best_antibody
)

print("\nAIS OPTIMIZATION COMPLETED")
print("Best parameters:", best_params)
print("Best CV RMSE:", best_fitness)

# ============================================================
# 10. SAVE OPTIMIZATION HISTORY
# ============================================================

history_df = pd.DataFrame(
    history_records
)

history_df.to_csv(
    BASE_DIR / "Ais_optimization_history.csv",
    index=False
)

# ============================================================
# 11. AIS CONVERGENCE GRAPH
# ============================================================

plt.figure(figsize=(11, 6))

plt.plot(
    history_df["Generation"],
    history_df["Global_Best_RMSE"],
    marker="o",
    linewidth=2,
    color="#2563EB",
    label="Global Best CV RMSE"
)

plt.plot(
    history_df["Generation"],
    history_df["Generation_Best_RMSE"],
    linestyle="--",
    color="#10B981",
    label="Generation Best CV RMSE"
)

plt.title(
    "AIS Clonal Selection Optimization Convergence",
    fontweight="bold"
)

plt.xlabel("Generation")
plt.ylabel("Cross-Validation RMSE")
plt.legend()
plt.tight_layout()

plt.savefig(
    BASE_DIR / "Ais_convergence_graph.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# ============================================================
# 12. TRAIN BASELINE AND AIS MODELS
# ============================================================

print("\nTraining baseline Random Forest...")

baseline_model = Pipeline([
    (
        "preprocessor",
        create_preprocessor()
    ),
    (
        "regressor",
        RandomForestRegressor(
            n_estimators=150,
            max_depth=20,
            min_samples_leaf=2,
            random_state=SEED,
            n_jobs=-1
        )
    )
])

baseline_model.fit(
    X_train_full,
    y_train_full
)

print("Training AIS-optimized Random Forest...")

ais_model = Pipeline([
    (
        "preprocessor",
        create_preprocessor()
    ),
    (
        "regressor",
        RandomForestRegressor(
            **best_params,
            random_state=SEED,
            n_jobs=-1
        )
    )
])

ais_model.fit(
    X_train_full,
    y_train_full
)

baseline_predictions = np.maximum(
    baseline_model.predict(X_test),
    0
)

ais_predictions = np.maximum(
    ais_model.predict(X_test),
    0
)

# ============================================================
# 13. EVALUATION METRICS
# ============================================================

def calculate_metrics(y_true, y_pred):

    y_true = np.asarray(
        y_true,
        dtype=float
    )

    y_pred = np.asarray(
        y_pred,
        dtype=float
    )

    mae = mean_absolute_error(
        y_true,
        y_pred
    )

    rmse = np.sqrt(
        mean_squared_error(
            y_true,
            y_pred
        )
    )

    r2 = r2_score(
        y_true,
        y_pred
    )

    valid = y_true != 0

    mape = (
        np.mean(
            np.abs(
                (y_true[valid] - y_pred[valid])
                / y_true[valid]
            )
        ) * 100
        if valid.any()
        else np.nan
    )

    return {
        "MAE": float(mae),
        "RMSE": float(rmse),
        "R2": float(r2),
        "MAPE": float(mape)
    }


baseline_metrics = calculate_metrics(
    y_test,
    baseline_predictions
)

ais_metrics = calculate_metrics(
    y_test,
    ais_predictions
)

metrics = {
    "Baseline_RF": baseline_metrics,
    "AIS_Optimized_RF": ais_metrics
}

metrics_df = pd.DataFrame(
    metrics
).T

print("\nMODEL COMPARISON")
print(metrics_df.round(4))

metrics_df.to_csv(
    BASE_DIR / "Ais_model_comparison.csv"
)

with open(
    BASE_DIR / "Ais_metrics.json",
    "w",
    encoding="utf-8"
) as file:

    json.dump(
        {
            "models": metrics,
            "best_ais_parameters": best_params,
            "best_cv_rmse": best_fitness,
            "optimization_algorithm": "CLONALG",
            "evaluation": (
                "Held-out same-date market records"
            )
        },
        file,
        indent=4
    )

# ============================================================
# 14. AIS ACCURACY GRAPH (R2)
# ============================================================

plt.figure(figsize=(10, 6))

names = [
    "Baseline RF",
    "AIS-Optimized RF"
]

values = [
    baseline_metrics["R2"],
    ais_metrics["R2"]
]

bars = plt.bar(
    names,
    values,
    color=["#94A3B8", "#2563EB"]
)

plt.axhline(
    0,
    color="black",
    linewidth=0.8
)

for bar, value in zip(bars, values):

    plt.annotate(
        f"{value:.4f}",
        (
            bar.get_x()
            + bar.get_width() / 2,
            bar.get_height()
        ),
        xytext=(
            0,
            5 if value >= 0 else -15
        ),
        textcoords="offset points",
        ha="center"
    )

plt.title(
    "AIS Model Performance (R² Score)",
    fontweight="bold"
)

plt.ylabel("R² Score")
plt.tight_layout()

plt.savefig(
    BASE_DIR / "Ais_accuracy_graph.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# ============================================================
# 15. AIS CORRELATION HEATMAP
# ============================================================

heatmap_data = df[
    [
        "Min_Price",
        "Max_Price",
        "Modal_Price"
    ]
].copy()

heatmap_data["Price_Spread"] = (
    heatmap_data["Max_Price"]
    - heatmap_data["Min_Price"]
)

correlation = heatmap_data.corr()

plt.figure(figsize=(9, 7))

sns.heatmap(
    correlation,
    annot=True,
    cmap="coolwarm",
    fmt=".3f",
    linewidths=0.5,
    square=True
)

plt.title(
    "AIS Agricultural Commodity Price Correlation",
    fontweight="bold"
)

plt.tight_layout()

plt.savefig(
    BASE_DIR / "Ais_heatmap.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# ============================================================
# 16. AIS MODEL COMPARISON GRAPH
# ============================================================

fig, axes = plt.subplots(
    2, 2,
    figsize=(14, 10)
)

metric_names = [
    "MAE",
    "RMSE",
    "R2",
    "MAPE"
]

for ax, metric in zip(
    axes.flatten(),
    metric_names
):

    values = [
        baseline_metrics[metric],
        ais_metrics[metric]
    ]

    bars = ax.bar(
        names,
        values,
        color=["#94A3B8", "#2563EB"]
    )

    ax.set_title(
        f"{metric} Comparison",
        fontweight="bold"
    )

    ax.set_ylabel(metric)

    for bar, value in zip(bars, values):

        ax.annotate(
            f"{value:.3f}",
            (
                bar.get_x()
                + bar.get_width() / 2,
                bar.get_height()
            ),
            xytext=(
                0,
                5 if value >= 0 else -15
            ),
            textcoords="offset points",
            ha="center",
            fontsize=9
        )

plt.suptitle(
    "Baseline Random Forest vs AIS-Optimized Random Forest",
    fontsize=16,
    fontweight="bold"
)

plt.tight_layout()

plt.savefig(
    BASE_DIR / "Ais_comparison_graph.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# ============================================================
# 17. AIS RESULT CSV
# ============================================================

result_df = X_test.copy()

result_df["Actual_Price"] = (
    y_test.to_numpy()
)

result_df["Baseline_Prediction"] = (
    baseline_predictions
)

result_df["AIS_Prediction"] = (
    ais_predictions
)

result_df["AIS_Absolute_Error"] = np.abs(
    result_df["Actual_Price"]
    - result_df["AIS_Prediction"]
)

result_df["AIS_Percentage_Error"] = (
    result_df["AIS_Absolute_Error"]
    / result_df["Actual_Price"]
    * 100
)

result_df = result_df.reset_index(
    drop=True
)

result_df.to_csv(
    BASE_DIR / "Ais_result.csv",
    index=False
)

print("\nSaved Ais_result.csv")

# ============================================================
# 18. AIS RESULT GRAPH
# ============================================================

plt.figure(figsize=(10, 8))

plt.scatter(
    result_df["Actual_Price"],
    result_df["AIS_Prediction"],
    alpha=0.5,
    s=22,
    color="#2563EB",
    label="AIS Predictions"
)

min_value = min(
    result_df["Actual_Price"].min(),
    result_df["AIS_Prediction"].min()
)

max_value = max(
    result_df["Actual_Price"].max(),
    result_df["AIS_Prediction"].max()
)

plt.plot(
    [min_value, max_value],
    [min_value, max_value],
    "--",
    color="red",
    linewidth=2,
    label="Perfect Prediction"
)

plt.xlabel("Actual Modal Price")
plt.ylabel("AIS Predicted Modal Price")

plt.title(
    "AIS Actual vs Predicted Commodity Prices",
    fontweight="bold"
)

plt.legend()
plt.tight_layout()

plt.savefig(
    BASE_DIR / "Ais_result_graph.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# ============================================================
# 19. AIS PREDICTION CSV
# ============================================================

prediction_df = X_test.copy()

prediction_df["Actual_Price"] = (
    y_test.to_numpy()
)

prediction_df["Baseline_Prediction"] = (
    baseline_predictions
)

prediction_df["AIS_Prediction"] = (
    ais_predictions
)

prediction_df = prediction_df.reset_index(
    drop=True
)

prediction_df.to_csv(
    BASE_DIR / "Ais_prediction.csv",
    index=False
)

print("\nSaved Ais_prediction.csv")

# ============================================================
# 20. AIS PREDICTION GRAPH
# ============================================================

plot_df = prediction_df.sample(
    n=min(100, len(prediction_df)),
    random_state=SEED
).sort_values(
    "Actual_Price"
).reset_index(drop=True)

plt.figure(figsize=(15, 7))

plt.plot(
    plot_df.index,
    plot_df["Actual_Price"],
    label="Actual Price",
    color="black",
    linewidth=2.5
)

plt.plot(
    plot_df.index,
    plot_df["Baseline_Prediction"],
    label="Baseline RF",
    color="#94A3B8",
    linestyle="--"
)

plt.plot(
    plot_df.index,
    plot_df["AIS_Prediction"],
    label="AIS-Optimized RF",
    color="#2563EB",
    linewidth=2
)

plt.xlabel("Sorted Test Sample Index")
plt.ylabel("Modal Price")

plt.title(
    "AIS Commodity Price Prediction Comparison",
    fontweight="bold"
)

plt.legend()
plt.tight_layout()

plt.savefig(
    BASE_DIR / "Ais_prediction_graph.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()

# ============================================================
# 21. AIS MARKET RECOMMENDATIONS
# ============================================================

# Recommendations based on reported modal prices.
# This is a gross-price ranking, not a profit ranking.

market_df = (
    df.groupby(
        [
            "State",
            "District",
            "Market",
            "Commodity",
            "Variety",
            "Grade"
        ],
        as_index=False
    )
    .agg(
        Average_Modal_Price=(
            "Modal_Price",
            "mean"
        ),
        Min_Reported_Price=(
            "Min_Price",
            "min"
        ),
        Max_Reported_Price=(
            "Max_Price",
            "max"
        ),
        Records=(
            "Modal_Price",
            "size"
        )
    )
)

market_df["Market_Rank"] = (
    market_df.groupby(
        [
            "Commodity",
            "Variety",
            "Grade"
        ]
    )["Average_Modal_Price"]
    .rank(
        ascending=False,
        method="dense"
    )
)

market_df = market_df.sort_values(
    [
        "Commodity",
        "Variety",
        "Grade",
        "Market_Rank"
    ]
)

market_df.to_csv(
    BASE_DIR / "Ais_market_recommendations.csv",
    index=False
)

# ============================================================
# 22. AIS MARKET RECOMMENDATION FUNCTION
# ============================================================

def ais_recommend_markets(
    commodity,
    variety=None,
    grade=None,
    top_n=5
):

    filtered = market_df[
        market_df["Commodity"].str.lower()
        == commodity.lower()
    ].copy()

    if variety is not None:

        filtered = filtered[
            filtered["Variety"].str.lower()
            == variety.lower()
        ]

    if grade is not None:

        filtered = filtered[
            filtered["Grade"].str.lower()
            == grade.lower()
        ]

    filtered = filtered.sort_values(
        "Average_Modal_Price",
        ascending=False
    )

    return filtered.head(top_n)


sample_commodity = df["Commodity"].iloc[0]

print("\nSample commodity:", sample_commodity)

print(
    ais_recommend_markets(
        sample_commodity,
        top_n=5
    ).to_string(index=False)
)

# ============================================================
# 23. SAVE AIS MODEL AS PKL
# ============================================================

joblib.dump(
    ais_model,
    BASE_DIR / "Ais_model.pkl"
)

print("\nSaved Ais_model.pkl")

# ============================================================
# 24. SAVE AIS METADATA AS H5
# ============================================================

# H5 stores AIS experiment metadata.
# The trained sklearn model is in Ais_model.pkl.

with h5py.File(
    BASE_DIR / "Ais_model.h5",
    "w"
) as h5:

    h5.attrs["algorithm"] = (
        "Artificial Immune System - CLONALG"
    )

    h5.attrs["predictor"] = (
        "RandomForestRegressor"
    )

    h5.attrs["model_file"] = (
        "Ais_model.pkl"
    )

    h5.attrs["best_cv_rmse"] = (
        best_fitness
    )

    params_group = h5.create_group(
        "best_parameters"
    )

    for key, value in best_params.items():
        params_group.attrs[key] = value

    h5.create_dataset(
        "convergence_rmse",
        data=history_df[
            "Global_Best_RMSE"
        ].to_numpy()
    )

print("Saved Ais_model.h5")

# ============================================================
# 25. SAVE AIS YAML CONFIGURATION
# ============================================================

config = {
    "project": (
        "AIS-Based Agricultural Commodity "
        "Price Prediction"
    ),
    "dataset": str(DATASET_PATH),
    "algorithm": "CLONALG",
    "predictor": "RandomForestRegressor",
    "target": TARGET,
    "features": FEATURES,
    "seed": SEED,
    "ais": {
        "population_size": POPULATION_SIZE,
        "generations": GENERATIONS,
        "top_antibodies": TOP_ANTIBODIES,
        "clones_per_antibody": (
            CLONES_PER_ANTIBODY
        ),
        "random_new_antibodies": (
            RANDOM_NEW_ANTIBODIES
        ),
        "cv_folds": CV_FOLDS
    },
    "best_parameters": best_params,
    "best_cv_rmse": float(best_fitness)
}

with open(
    BASE_DIR / "Ais_config.yaml",
    "w",
    encoding="utf-8"
) as file:

    yaml.safe_dump(
        config,
        file,
        sort_keys=False
    )

print("Saved Ais_config.yaml")

# ============================================================
# 26. SAVE AIS PROJECT SUMMARY
# ============================================================

summary = {
    "dataset_records": int(len(df)),
    "unique_states": int(
        df["State"].nunique()
    ),
    "unique_markets": int(
        df["Market"].nunique()
    ),
    "unique_commodities": int(
        df["Commodity"].nunique()
    ),
    "unique_dates": int(
        df["Arrival_Date"].nunique()
    ),
    "algorithm": "CLONALG",
    "best_parameters": best_params,
    "baseline_metrics": baseline_metrics,
    "ais_metrics": ais_metrics,
    "best_cv_rmse": float(best_fitness),
    "evaluation_note": (
        "Test metrics evaluate held-out market "
        "records from the available dataset, "
        "not future-date forecasting."
    )
}

with open(
    BASE_DIR / "Ais_project_summary.json",
    "w",
    encoding="utf-8"
) as file:

    json.dump(
        summary,
        file,
        indent=4
    )

# ============================================================
# 27. AIS IMPROVEMENT ANALYSIS
# ============================================================

baseline_rmse = baseline_metrics["RMSE"]
ais_rmse = ais_metrics["RMSE"]

improvement = (
    (baseline_rmse - ais_rmse)
    / baseline_rmse
) * 100 if baseline_rmse > 0 else np.nan

print("\n" + "=" * 70)
print("AIS OPTIMIZATION RESULTS")
print("=" * 70)

print("\nBaseline RMSE:", baseline_rmse)
print("AIS RMSE:", ais_rmse)

print(
    "AIS RMSE Improvement (%):",
    round(improvement, 4)
)

if improvement > 0:
    print("AIS improved test RMSE.")
elif improvement < 0:
    print("AIS did not improve test RMSE.")
else:
    print("AIS and baseline RMSE are equal.")

# ============================================================
# 28. FINAL FILE LIST
# ============================================================

print("\n" + "=" * 70)
print("ALL AIS RESULTS GENERATED")
print("=" * 70)

for file in sorted(BASE_DIR.glob("Ais_*")):

    if file.is_file():
        print(" -", file.name)

print("\nOutput directory:")
print(BASE_DIR)

print("\nAIS PROJECT COMPLETED SUCCESSFULLY")
```
